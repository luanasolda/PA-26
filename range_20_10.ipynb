```python
dia = "segunda"

match dia:

  case "segunda":
    print("Dia de faltar")
  case "terça":
    print("Aula da Leticia")
  case "quarta":
    print("matadade da semana")
  case "quinta":
    print("perto de sexta")
  case "sexta":
    print("só felicidade")
  case "sabado":
    print("enquanto não chega domingo")
  case "domingo":
    print("tristeza")
  case _:
    print("opção invalida")
```

**Exercicios Range**

```python
# 1. Mostrar os números de 0 até 9
for numero in range(10):
  print(numero)
```

```python
# 2. Mostrar os números de 1 até 10

for numero in range(1,11):
  print(numero)
```

```python
# 3. Mostrar os números de 5 até 15
for numero in range(5,16):
  print(numero)
```

```python
# 4. Mostrar os números pares de 0 até 20
for numero in range(21):
  print(numero)
```

```python
# 5. Mostrar os números ímpares de 1 até 15
for numero in range(1,16):
  print(numero)
```

```python
# 6. Fazer uma contagem regressiva de 10 até 1
for numero in range(10,0,-1):
  print(numero)
```

```python
# 7. Mostrar a tabuada do número 3
for numero in range(3,33,3):
  print(numero)
```

**Exercicos Match**

```python
# 1. Verificar a cor
cor = "azul"

match cor:

  case "azul":
    print("Cor azul")

  case _:
    print("outra cor")
```

```python
# 2. Verificar o número
numero = "dois"

match numero:

  case "um":
    print("Você digitou um")
  case "dois":
    print("Você digitou dois")
  case "três":
    print("Você digitou três")
  case _:
    print("opão invalida")
```

```python
# 3. Menu de bebidas

bebida = 2

match bebida:

  case 1:
    print("água")
  case 2:
    print("suco")
  case 3:
    print("refrigerante")
  case _:
    print("opção invalida")
```

```python
# 4. Verificar o final de semana

dia = "segunda"

match dia:

  case "segunda":
    print("Dia de semana")
  case "terça":
    print("Dia de semana")
  case "quarta":
    print("Dia de semana")
  case "quinta":
    print("Dia de semana")
  case "sexta":
    print("Dia de semana")
  case "sabado":
    print("Final de semana")
  case "domingo":
    print("Final de semana")
  case _:
    print("opção invalida")
```

```python
# . Verificar a nota

nota = "A"

match nota:

  case "A":
    print("Excelente")
  case "B":
    print("Bom")
  case "C":
    print("Precisa melhorar")
  case _:
    print("Opção invalida")
```
