```python
# exercicio 3
# Receba dois números e mostre qual é o maior. Se os dois números forem iguais, diga que
# eles são iguais

a = int(input("Digite o primeiro número: "))
b = int(input("Digite o segundo número: "))

if a > b:
  print(f"{a} é meior que {b}")

elif b > a:
  print(f"{b} é maior que {a}")

else:
  print(f"Os dois números são iguais!")
```

```python
# exercicio 4
# Receba a nota de um aluno e mostre sua situação:
# Nota maior ou igual a 7 → Aprovado
# Nota maior ou igual a 5 e menor que 7 → Recuperação
# Nota menor que 5 → Reprovado


a = float(input(f"Digite sua primeira nota: "))
b = float(input(f"Digite sua segunda nota: "))
c = float(input(f"Digite sua terceira nota: "))


media = (a + b + c) / 3

if media >= 7:
  print(f" Sua média é de {media} e você foi aprovado!")

elif media >= 5 and media < 7:
  print(f"Sua média é de {media} e você ficou de recuperação")

elif media < 5:
  print(f"Sua média foi de {media} e você foi reprovado!")
```

```python
# exercicio 5
# Receba o valor de uma compra. Se o valor for maior ou igual a 100, aplique 10% de
# desconto e mostre o valor final. Caso contrário, mostre o valor normal da compra

a = float(input("Insira o valor da compra: "))

if a >= 100:
  des = a * 0.10
  val = a - des
  print(f"Você gastou R${a}, e recebeu um desconto de 10%. Valor a ser pago: R${val}")

else:
  print(f"Valor a ser pago: R${a}")
```

```python
# exercicio 6
# Receba um número e diga se ele é positivo, negativo ou zero.

a = float(input("Digite um número negativo ou positivo: "))

if a > 0:
  print(f"O número {a} é positivo!")

elif a < 0:
  print(f"O número {a} é negativo!")

else:
  print(f"O número é zero")
```

```python
# exercicio 7
# Receba a idade de uma pessoa e classifique-a de acordo com sua faixa etária:
# Menor de 12 anos → Criança
# De 12 a 17 anos → Adolescente
# De 18 a 59 anos → Adulto
# 60 anos ou mais → Idoso

a = int(input("Digite sua idade: "))

if a < 12:
  print("Você ainda é uma criança!")

elif a > 12 and a < 17:
  print("Você é um adolecente!")

elif a > 18 and a < 59:
  print("Você é um adulto!")

else:
  print("Você é um idoso!")
```

```python
# exercicio 8
# Receba duas notas, calcule a média e diga se o aluno foi aprovado ou reprovado. O aluno
# será aprovado se a média for maior ou igual a 7.

a = float(input("Digite sua primeira nota: "))
b = float(input("Digite sua segunda nota: "))

media = (a + b) / 2

if media >= 7:
  print(f"Sua média é {media} e você foi aprovado!")

else:
  print(f"Sua média é {media} e você foi reprovado!")
```

```python
# exercicio 10
# Receba o peso e a altura de uma pessoa, calcule o IMC usando a fórmula:
# IMC = peso / (altura × altura)
# Depois, classifique o resultado:

# IMC menor que 18.5 → Abaixo do peso
# IMC de 18.5 a 24.9 → Peso normal
# IMC de 25 a 29.9 → Sobrepeso
# IMC maior ou igual a 30 → Obesidade

a = float(input("Digite sua altura: "))
b = float(input("Digite seu peso: "))

imc = b / (a * a)

print("")
print(f"Seu IMC é {imc}")
print("")

if imc < 18.5:
  print("Você esta abaixo do peso.")

elif imc > 18.5 and imc < 24.9:
  print("Você esta com peso normal.")

elif imc > 25 and imc < 29.9:
  print("Você esta sobrepeso.")

else:
  print("Você esta obeso")
```
